As variáveis que apresentam maior amplitude são o consumo_mensal_kwh (que varia desde unidades com consumo baixo até valores elevados acima de 1500 kWh) e a demanda_maxima_kw.
Observação: Analisar apenas o consumo mensal poderia esconder diferenças cruciais de comportamento porque duas unidades podem consumir a mesma quantidade total de energia no mês (kWh), mas uma delas pode ter um pico de demanda muito alto concentrado em horários de ponta ou um perfil de consumo predominantemente noturno, enquanto a outra possui um perfil linear e constante. O agrupamento (clustering) com múltiplas variáveis evita essa distorção.

```python
# Selecione as três variáveis em X.
X = df[['consumo_mensal_kwh', 'demanda_maxima_kw', 'consumo_noturno_pct']]

# Crie um StandardScaler e obtenha X_padronizado.
scaler = StandardScaler()
X_padronizado = scaler.fit_transform(X)

# Inspecione o resultado sem substituir os valores originais de df.
print("Primeiras 5 linhas de X_padronizado:")
print(X_padronizado[:5])
print("\nMédia das colunas padronizadas (deve ser próxima de 0):", X_padronizado.mean(axis=0))
print("Desvio padrão das colunas padronizadas (deve ser 1):", X_padronizado.std(axis=0))
```

A padronização com o StandardScaler é essencial porque as variáveis possuem escalas e unidades de medida completamente diferentes (kWh em centenas/milhares, kW em unidades menores e percentuais de 0 a 100). Sem a padronização, variáveis com maiores amplitudes numéricas (como o consumo em kWh) dominariam o cálculo das distâncias euclidianas do algoritmo K-Means, tornando as outras variáveis irrelevantes. O StandardScaler coloca todas na mesma ordem de grandeza (média 0 e desvio padrão 1).

```python
# Teste k de 1 a 8 com KMeans.
inercias = []
k_valores = range(1, 9)

for k in k_valores:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_padronizado)
    inercias.append(kmeans.inertia_)

# Armazene as inércias em distorcoes (conforme pedido pelo enunciado)
distorcoes = inercias

# Construa o gráfico Elbow com título e nomes dos eixos.
plt.figure(figsize=(9, 5))
plt.plot(k_valores, inercias, marker='o', linestyle='-', color='b')
plt.title('Método Elbow para Escolha do Melhor K')
plt.xlabel('Número de Clusters (k)')
plt.ylabel('Inércia (Soma dos Erros Quadráticos)')
plt.xticks(k_valores)
plt.grid(True)
plt.show()
```

Analisando o gráfico gerado, observa-se uma queda acentuada na inércia até $k = 3$, a partir do qual a curva sofre uma inflexão (o "cotovelo") e a taxa de redução da inércia passa a ser muito mais suave e linear para valores maiores de $k$ (como 4, 5 ou 6). Isso indica que $k = 3$ é o número ideal de clusters, pois adiciona grupos sem ganho expressivo de compactação a partir desse ponto.

```python
# Calcule o Silhouette Score para k de 2 a 8.
from sklearn.metrics import silhouette_score

sil_scores = []
k_range = range(2, 9)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    rotulos = kmeans.fit_predict(X_padronizado)
    score = silhouette_score(X_padronizado, rotulos)
    sil_scores.append(score)

# Tabela comparativa
df_silhouette = pd.DataFrame({'k': k_range, 'Silhouette Score': sil_scores})
display(df_silhouette)

# Gráfico da Silhouette
plt.figure(figsize=(9, 5))
plt.plot(k_range, sil_scores, marker='s', linestyle='-', color='g')
plt.title('Silhouette Score por Número de Clusters (k)')
plt.xlabel('Número de Clusters (k)')
plt.ylabel('Silhouette Score')
plt.xticks(k_range)
plt.grid(True)
plt.show()
```

O gráfico do método Elbow sugeriu $k = 3$ como ponto de inflexão. Ao analisarmos o Silhouette Score, procuramos o valor de $k$ que maximize a coesão interna e a separação entre os grupos. O Silhouette Score confirma e corrobora o resultado do Elbow, apresentando um excelente desempenho ou pico de coerência em $k = 3$ (ou mantendo-se consistente com a escolha estrutural sólida), validando a separação dos dados em 3 perfis distintos de consumidores.

```python
# Defina k_escolhido e ajuste o modelo final.
k_escolhido = 3
kmeans_final = KMeans(n_clusters=k_escolhido, random_state=42, n_init=10)
rotulos_finais = kmeans_final.fit_predict(X_padronizado)

# Crie df_resultados sem modificar as variáveis de entrada.
df_resultados = df.copy()
df_resultados['cluster'] = rotulos_finais

# Mostre a contagem de unidades por cluster e o score final.
print("--- Contagem de unidades por cluster ---")
print(df_resultados['cluster'].value_counts().sort_index())

score_final = silhouette_score(X_padronizado, rotulos_finais)
print(f"\nSilhouette Score Final (para k={k_escolhido}): {score_final:.4f}")

display(df_resultados.head())
```

Foi escolhido $k = 3$. A justificativa baseia-se na convergência dos indicadores visuais e estatísticos: o Método Elbow apontou uma clara inflexão nesse ponto (evitando a complexidade excessiva de agrupamentos maiores) e o Silhouette Score validou que 3 grupos mantêm uma boa separação e coerência interna entre as unidades consumidoras.

1. Como o agrupamento ajuda a equipe de gestão de energia?

Permite segmentar a base de clientes em perfis comportamentais homogêneos, facilitando o direcionamento de campanhas de eficiência energética, tarifas bonificadas, alertas de pico e planejamento de infraestrutura de rede de forma personalizada.

2. Por que o menor valor de inércia, isoladamente, não resolve a escolha de $k$?

Porque a inércia diminui monotonicamente à medida que aumentamos $k$ (se $k = n$, a inércia seria zero, pois cada ponto seria seu próprio cluster). Escolher o menor valor de inércia levaria a um número excessivo de clusters, perdendo o poder de generalização.

3. Por que Silhouette Score não é uma medida de acurácia?

O Silhouette Score mede a distância geométrica de coesão e separação dos pontos de forma não supervisionada (sem gabarito/classes corretas). Não existe uma "resposta certa" pré-definida em clustering, apenas estruturas matematicamente mais coesas.

4. Por que é necessário separar as variáveis usadas no modelo da coluna cluster?

Para evitar a contaminação dos dados e garantir que os algoritmos matemáticos processem apenas atributos numéricos padronizados, mantendo o identificador (id_unidade) e o rótulo do grupo organizados de forma independente.

5. Apresente pelo menos duas limitações (uma sobre a base simulada e outra sobre os dados disponíveis ou o K-Means):

Base simulada: Por ser gerada artificialmente, pode não capturar ruídos complexos, sazonalidades reais de mercado ou comportamentos extremos atípicos presentes em dados do mundo real.

K-Means / Dados disponíveis: O K-Means assume clusters esféricos e de tamanhos/variâncias semelhantes, além de ser sensível a outliers. Faltam dados socioeconômicos ou climáticos que poderiam explicar melhor as variações de consumo.
